<span style="color: #177a7d; font-size: 20px;">En aquest notebook veurem què és una API REST, com funciona i com podem utilitzar-la per obtenir dades d'una pàgina web.</span>

<span style="color: #4aead5; font-size:14px">**Què és una API REST?** Com s'utilitza? Per què és tan important en el món de les dades? </span>

<span style="font-size:14px">Una API (Interfície de Programació d'Aplicacions) és un conjunt de regles i protocols que permeten que diferents programes es comuniquin entre si. REST (Representational State Transfer) és un estil d'arquitectura per a la construcció de serveis web que utilitza els mètodes HTTP per realitzar operacions sobre recursos.</span>

<span style="font-size:14px">Una API RESTful és una API que segueix els principis de REST. Això significa que utilitza els mètodes HTTP (GET, POST, PUT, DELETE) per realitzar operacions sobre recursos, i que els recursos es representen com a URL. Per exemple, si volem obtenir informació sobre un usuari amb ID 123, podríem fer una petició GET a l'URL `https://api.example.com/users/123`. Aquesta petició retornaria la informació de l'usuari en format JSON o XML, depenent de com estigui configurada l'API.</span>

<span style="color: rgb(14, 111, 111); font-size:14px">**Com s'utilitza?** </span>

1. <span style="font-size:14px">Primer, necessitem conèixer l'URL de l'API i els endpoints disponibles. Un endpoint és una URL específica que representa un recurs o una acció que podem realitzar. Per exemple, `https://api.example.com/users` podria ser un endpoint per obtenir la llista d'usuaris.</span> <br>
2. <span style="font-size:14px">A continuació, podem utilitzar una eina com Postman o cURL per realitzar peticions a l'API. Per exemple, si volem obtenir la llista d'usuaris, podríem fer una petició **GET** a `https://api.example.com/users` i rebríem una resposta amb la informació dels usuaris.</span> <span style="font_size:14px">**GET** `https://api.example.com/users` <br>
3. <span style="font-size:14px">Si tot va bé, rebriem els codis d'estat HTTP 200 (OK) i la resposta amb les dades sol·licitades. Si hi ha un error, podríem rebre codis d'estat com 404 (Not Found) o 500 (Internal Server Error).</span><br>
4. <span style="font-size:14px">Podem utilitzar els mètodes POST, PUT i DELETE per crear, actualitzar o eliminar recursos respectivament. Per exemple, si volem crear un nou usuari, podríem fer una petició **POST** a `https://api.example.com/users` amb les dades del nou usuari en el cos de la petició.</span> **POST** `https://api.example.com/users`<br>

<span style="color: rgb(111, 40, 14)">---------------------------------------------------- **Exemple de codi en Python per fer una petició GET a una API REST:** ---------------------------------------------------------------------- </span>
```python
import requests

response = requests.get('https://api.example.com/users')
if response.status_code == 200:
    data = response.json()
    print(data)
else:
    print(f'Error: {response.status_code}')
```

<span style="font-size:14px">Les API REST són essencials en el món de les dades perquè permeten accedir a grans quantitats d'informació de manera estructurada i eficient. Moltes empreses i organitzacions ofereixen les seves dades a través d'API REST, cosa que facilita als analistes de dades obtenir informació valuosa per als seus anàlisis.</span>
<span style="font-size:14px">A més, les API REST permeten integrar dades de diferents fonts i crear aplicacions que poden utilitzar aquestes dades de manera dinàmica. Per exemple, una aplicació de predicció del temps podria utilitzar una API REST per obtenir dades meteorològiques en temps real i proporcionar prediccions als usuaris.</span>
<span style="font-size:14px">En resum, les API REST són una eina poderosa per accedir i utilitzar dades en el món digital, i són una part essencial de l'ecosistema de dades actual.</span>

````python
#Llista de publicacions (/posts):
publicacions_url='https://jsonplaceholder.typicode.com/posts'
response=requests.get(publicacions_url)
print(f"Codi d'estat: {response.status_code}")    #Com indica, el mètode .status_code ens indica el codi de l'estat de la consulta: https://www.siteground.es/kb/codigos-error-http-explicados/
````

````python
import json 
dades=json.loads(response.text)    #El mètode .text ens retorna el contingut de la resposta en format text, i json.loads() converteix aquest text en un objecte de Python (en aquest cas, una llista de diccionaris).
print(f"Nombre de publicacions: {len(dades)}")    #Mostrem el nombre de publicacions recuperades, que hauria de ser 100 segons l'API.
````

````python
#Si volem el header de la resposta:
import pprint
pprint.pprint(f"Headers de la resposta: {response.headers}")
````

````python
#Si volem mostrar les 5 primeres publicacions:
for publicacio in dades[:5]:
    print(f"ID: {publicacio['id']}, Títol: {publicacio['title']}")
````

````python
#Llista d'usuaris:
usuaris='https://jsonplaceholder.typicode.com/users'
response_users=requests.get(usuaris)
print(f"Codi d'estat: {response_users.status_code}")

dades_users=json.loads(response_users.text)
print(f"Hi ha {len(dades_users)} usuaris en la teva consulta")

#Visualitzem l'estructura del document JSON:
pprint.pprint(dades_users[0])   #Mostrem el primer usuari per veure l'estructura del JSON.
````

````python
#Llista de tasques: 
todos='https://jsonplaceholder.typicode.com/todos'
response_tasques=requests.get(todos)
print(f"Codi d'estat: {response_tasques.status_code}")

dades_tasques=json.loads(response_tasques.text)
print(f"La teva consulta conté {len(dades_tasques)} dades")

pprint.pprint(dades_tasques[0])
````

1. <span style="font-size:14px">Utilitza l’API d’Open Data BCN per: </span>

   - <span style="font-size:14px">Cercar un dataset del teu interès mitjançant package_search i package_show.</span>

````python
import pandas as pd

BASE_URL = "https://opendata-ajuntament.barcelona.cat/data/api/3/action/"

# Recuperar tots els datasets amb detalls (paginació de 100 en 100)
tots_datasets = []
limit = 100
offset = 0

while True:
    params = {"limit": limit, "offset": offset}
    r = requests.get(BASE_URL + "current_package_list_with_resources", params=params)
    batch = r.json()["result"]
    if not batch:
        break
    tots_datasets.extend(batch)
    offset += limit

print(f"Total datasets trobats: {len(tots_datasets)}")
````

````python
# Convertir a DataFrame per visualitzar millor
df_datasets = pd.DataFrame([
    {
        "nom": d["name"],
        "titol": d["title"],
        "num_recursos": len(d["resources"]),
        "formats": ", ".join(set(r["format"] for r in d["resources"] if r["format"]))
    }
    for d in tots_datasets
])

df_datasets.head(20)   # mostra els primers 20 datasets amb el seu nom, títol, nombre de recursos i formats disponibles.
````

````python
# Treballem amb el dataset de Immigrants a Barcelona:
DATASET_NOM = "pad_imm_mdbas" 

r = requests.get(BASE_URL + "package_show", params={"id": DATASET_NOM})
dataset = r.json()["result"]

print(f"Títol: {dataset['title']}")
print(f"\nRecursos disponibles:")
for rec in dataset["resources"]:
    print(f"  Format: {rec['format']} | Nom: {rec['name']}")
    print(f"  resource_id: {rec['id']}\n")
````
