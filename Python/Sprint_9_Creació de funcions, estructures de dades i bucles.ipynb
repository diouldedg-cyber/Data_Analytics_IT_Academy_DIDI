### 1. Calculadora de l'índex de massa corporal

***Calcular IMC***

<span style="font-size: 14px;"> Para calcular el **IMC** es necesario *dividir* el **peso en kg** de un paciente entre el **cuadrado de su altura en metros**. <br> Con esto obtendremos un valor que se situará en uno de los siguientes rangos: <br> </span>
<span style="color: yellow; font-size: 12px;">**1. Bajo peso: menos de 18,5.** <br>
<span style="color: yellow; font-size: 12px;">**2. Peso normal: 18,5-24,9.** <br>
<span style="color: yellow; font-size: 12px;">**3. Sobrepeso: 25-29,9.** <br>
<span style="color: yellow; font-size: 12px;">**4. Obesidad: más de 30.** <br>


```python
# La funció ha de rebre dos parametres: el pes i l'alçada d'una persona, i ha de retornar el seu IMC (Índex de Massa Corporal).
def IMC (pes, alçada):
    if not isinstance(pes, (int, float)) or not isinstance(alçada, (float)): #isinstance --> isinstance(obj, tipus) comprova si obj és d'un tipus específic i retorna True o False.
        raise TypeError(f"Has introduït un tipus de valor no valid")
    if alçada <= 0:
        raise ValueError(f"L'alçada ha de ser un valor positiu: {alçada}")    # raise --> ALERTA! Error greu! Para tot i gestiona'm!
    if alçada <0.30 or alçada >3.00:
        raise ValueError (f"L'alçada ha de estar en metres, entre 0.30-3.00 m. Alçada introduïda: {alçada}")
    
    IMC_persona=pes/(alçada**2)
    
    if IMC_persona <= 18.5:
        resultat="Pes baix"
    elif 18.5 < IMC_persona <=24.9:
        resultat="Pes normal"
    elif 25 <= IMC_persona <=29.9:
        resultat="Sobrepes"
    else:
        resultat="Obesitat"
    return IMC_persona, resultat
```

```python
# --- CRIDEM LA FUNCIÓ ---
pes = float(input("Introdueix el teu pes en kg: "))     
alçada = float(input("Introdueix la teva alçada en metres: "))
IMC_persona, resultat = IMC(pes, alçada)
print(f"El teu IMC és: {IMC_persona:.2f}. Resultat: {resultat}") 
```

### 2.	Convertidor de temperatures
<span style="font-size: 14px;">Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són **Celsius (°C)**, **Fahrenheit (°F)** i **Kelvin (K)**. També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re). </span>


<span style="font-size: 11.5px;">De Celsius a Fahrenheit:<br>
°F = (°C × 9/5) + 32 <br>
De Fahrenheit a Celsius:<br>
°C = (°F – 32) × 5/9 <br> 
De Celsius a Kelvin: <br>
K = °C + 273.15 <br>
De Kelvin a Celsius:<br>
°C = K – 273.15 </span>

```python
def Temperature_conversor (temp):
    Celsius= {"C": temp,
              "F": (temp*9/5)+32,
              "K": temp+273.15,
              "Re": temp*0.8,
              "Ra": (temp+273.15)*1.8
            }
    return Celsius
```

```python
def conversor_temperaturas(t_origen, t_desti, temp):
    import string

    t_origen = string.capwords(t_origen)
    t_desti = string.capwords(t_desti)
    
    # Verificar que existan en el DataFrame
    if t_origen not in df.index or t_desti not in df.columns:
        return f"❌ '{t_origen}' -> '{t_desti}' no disponible"          #Per posar emojis: "Wnd + ." i copias l'emoji directament 
    
    formula = df.loc[t_origen, t_desti]
    
    if pd.isna(formula) or str(formula).strip() == '-':
        return "❌ Conversión no disponible"
    
    # Convertir fórmula a código Python
    formula_python = str(formula)
    formula_python = formula_python.replace('−', '-').replace('×', '*').replace('÷', '/')
    
    # Reemplazar variables de temperatura (°C, °F, K, etc.)
    vars_temp = {'C': temp, 'F': temp, 'K': temp, 'Ra': temp, 'Re': temp}
    for simbolo, valor in vars_temp.items():
        formula_python = formula_python.replace(simbolo, str(valor))
    
    try:
        resultado = eval(formula_python)
        return f"La teva conversió: ✅ {temp} {t_origen} = {round(resultado, 2)} {t_desti} 🌡️"
    except Exception as e:
        return f"❌ Error cálculo: {e}"
```

### 3. Comptador de paraules d'un text.
<span style= "font-size: 14px;"> **Escriu una funció que donat un text, mostri les vegades que apareix cada paraula**. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament. <br> </span>
<span style= "color: red; font-size: 16px;"> **(EXTRA)**: Quina és la ***llargada mitjana*** de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3 </span>


```python

def comp_paraules(txt):
    import string
    paraules = txt.lower().split() # Transformem i després separem 
    paraules = [p.strip(string.punctuation) for p in paraules] # strip() --> Elimina els caràcters especificats al principi i al final de la cadena. En aquest cas, elimina la puntuació.
    compt = len(paraules)
    llargada_total = sum(len(p) for p in paraules)
    mitjana = llargada_total / compt if compt > 0 else 0

    dicc_paraules = {}
    for paraula in paraules:
        if paraula in dicc_paraules:
            dicc_paraules[paraula] = dicc_paraules.get(paraula, 0) + 1
        else:
            dicc_paraules[paraula] = 1
    
    dicc_ordenat=dict(sorted(dicc_paraules.items(), key=lambda item: item[1], reverse=True)) #La funció sorted() ordena en ordre descendent per defecte. 

    return dicc_ordenat, compt, mitjana   
```

### 4. Diccionari invers (amb possibilitat de duplicats)

<span style="font-size:14px"> Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; <br> si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.


```python
def dicc_invers(d):
    # avisar si hi ha valors duplicats
    duplicats = {}
    for clau, valor in d.items():
        if valor in duplicats:                      #El if va aquí dalt perquè sinó, el valor es sobreescriuria i no podríem detectar que està duplicat. 
            print(f"El valor '{valor}' està duplicat. No es pot crear inversió única.")
        duplicats.setdefault(valor, []).append(clau) #setdefault() method returns the value of the item with the specified key.

    # invertir agrupant totes les claus amb el mateix valor
    dicc_invertit = {}
    for clau, valor in d.items():       #The items() method returns a view object. The view object contains the key-value pairs of the dictionary, as tuples in a list.
        if valor in dicc_invertit:      #dict_items([('a', 1), ('b', 2), ('c', 1), ('z', 'banana'), ('k', 'kiwi'), ('m', 'kiwi')])
            dicc_invertit[valor].append(clau)   # ja tenim llista, només afegim
        else:
            dicc_invertit[valor] = [clau]       # comencem la llista

    return dicc_invertit   
```

```python
dicc={'a':1, 'b':2, 'c':1, 'z':'banana', 'k':'kiwi', 'm':'kiwi'}
invers=dicc_invers(dicc)
print(invers)
```

### 1. Comptador i endreçador de paraules d'un text.

<span style="font-size:14px"> El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals <br> del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z.

```python
def comp_endreçador(txt):
    import re
    
    paraules = re.findall(r'\b[a-zà-ÿ]+\b', txt.lower()) #re.findall(pattern, string)-->retorna una llista de totes les coincidències del patró en la cadena. El patró r'\b[a-zà-ÿ]+\b' busca paraules completes, ignorando puntuació.
    compt = len(paraules)
    mitjana = sum(len(p) for p in paraules) / compt if compt else 0
    
    # Creem el diccionari per lletre
    dicc_per_lletra = {}
    for paraula in paraules:
        inicial = paraula[0]
        if inicial not in dicc_per_lletra:
            dicc_per_lletra[inicial] = {}
        dicc_per_lletra[inicial][paraula] = dicc_per_lletra[inicial].get(paraula, 0) + 1
    return dicc_per_lletra
```

<span style="font-size:14px"> En aquest cas s'ha treballat utilizant el mòdul ***random*** de la llibreria NumPy. Però es podia haver utilitzat el mòdul ***secret***. <br> Per quin motiu no s'ha fet servir ***secret***? 
El mòdul secrets <font color="magenta"> genera números aleatoris més aleatoris i millor distribuïts, però és menys eficient </font>; és excessiu per simulacions o proves, tot i que la gent sovint prioritza qualitat sobre velocitat. La biblioteca random usa tècniques senzilles i ràpides, mentre que secrets arriba a l'extrem d'uns aleatoris tan imprevisibles que no es poden explotar per atacar la seguretat.

```python
def gen_contrasenya(long, majus=True, minus=True, nume=True, esp=True):
    import random
    import string 
    
    caracters=''
    #Creem filtre longitud mínima requerida
    if long < 10:
        raise ValueError("La longitud de la contrasenya ha de ser almenys de 10 caràcters.")
    
    #Filtres de les característiques de la contrasenya
    if majus:
        caracters += string.ascii_uppercase
    if minus:
        caracters += string.ascii_lowercase
    if nume:
        caracters += string.digits
    if esp:
        caracters += string.punctuation
    if not caracters:
        raise ValueError("Almenys un tipus de caràcter ha de ser seleccionat.")
    while True: 
        contrasenya =''.join(random.choice(caracters) for _ in range(long)) 
        if (any(c.islower() for c in contrasenya) if min else True
            and any(c.isupper() for c in contrasenya) if maj else True
            and sum(c.isdigit() for c in contrasenya) >= 3 if num else True
            and any(c in string.punctuation for c in contrasenya) if esp else True):
            break                                                                             #Problema: es queda executant. Buscar la forma de fer que pari la iteració i continue. El continue no em funciona. 
        raise ValueError ("La contrasenya generada no cumpleix els requisits: majúsculas, minúsculas, números o caracters especials. Genra una nova contrasenya.")

    return contrasenya
```

#### Anàlisi de les dades del arxiu: 

1. <span style="font-size:12px; color:cyan"> **Definim la funció**

```python
def analisi_partits(txt):
    import re
    import pandas as pd

    txt=txt.read()

    dades = []
    for linia in txt.strip().split('\n'):
        patro = r"^(.+?)\s+(\d+-\d+)\s+(.+)$"
        coincidencies = re.findall(patro, linia)
        for match in coincidencies:
            local, gols, visitant = match
            gols_local, gols_visitant = map(int, gols.split('-'))
            dades.append({
                'local': local.strip(),
                'gols_local': gols_local,
                'gols_visitant': gols_visitant,
                'visitant': visitant.strip()
            })
    
    df_partits = pd.DataFrame(dades)
    
    # Total gols per equip
    #print("\nTotal gols per equip:")
    gols_per_equip = pd.concat([
        df_partits.groupby('local')['gols_local'].sum(),
        df_partits.groupby('visitant')['gols_visitant'].sum()
    ]).groupby(level=0).sum()
    #print(gols_per_equip)
    
    # Analisis estadístic complet per equip
    equips = pd.concat([df_partits['local'], df_partits['visitant']]).unique()
    estadistiques = []
    
    for equip in equips:
        # Locals
        locals = df_partits[df_partits['local'] == equip]
        guanyats_local = len(locals[locals['gols_local'] > locals['gols_visitant']])
        empatats_local = len(locals[locals['gols_local'] == locals['gols_visitant']])
        perduts_local = len(locals[locals['gols_local'] < locals['gols_visitant']])
        
        # Visitants
        visitants = df_partits[df_partits['visitant'] == equip]
        guanyats_visitant = len(visitants[visitants['gols_visitant'] > visitants['gols_local']])
        empatats_visitant = len(visitants[visitants['gols_visitant'] == visitants['gols_local']])
        perduts_visitant = len(visitants[visitants['gols_visitant'] < visitants['gols_local']])
        
        # Totals
        total_guanyats = guanyats_local + guanyats_visitant
        total_empatats = empatats_local + empatats_visitant
        total_perduts = perduts_local + perduts_visitant
        total_gols = locals['gols_local'].sum() + visitants['gols_visitant'].sum()
        total_encaixats = locals['gols_visitant'].sum() + visitants['gols_local'].sum()
        total_punts = total_guanyats * 3 + total_empatats
        
        estadistiques.append({
            'Equip': equip,
            'partits guanyats': total_guanyats,
            'partits empatats': total_empatats,
            'partits perduts': total_perduts,
            'gols': total_gols,
            'gols encaixats': total_encaixats,
            'punts': total_punts
        })
    
    df_classificacio = pd.DataFrame(estadistiques)
    df_classificacio = df_classificacio.sort_values('punts', ascending=False).reset_index(drop=True)
    
    mitjana_gols = (df_partits['gols_local'] + df_partits['gols_visitant']).mean()
    
    # FORMATEIG FINAL visual taula
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', None)


    display (df_classificacio)      #https://www.geeksforgeeks.org/python/how-to-pretty-print-an-entire-pandas-series-or-dataframe/

```
