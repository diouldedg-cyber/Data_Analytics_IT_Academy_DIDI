```python

def graficos_automaticos(df):
    import pandas as pd
    import matplotlib.pyplot as plt
    from pathlib import Path

    carpeta = Path(r"C:\Users\dioul\OneDrive\Escritorio\Fonaments D'anàlisi de dades (FADD)\Curs d'especialització Data Analys\Python\Sprint _10\Gràfics_automatitzats")
    carpeta.mkdir(exist_ok=True)

    for nom_columna in df.columns:
        serie = df[nom_columna]
        valors_unics = serie.dropna().unique()
        n_unics = len(valors_unics)

        plt.figure(figsize=(8, 6))

        if serie.dtype in ["int64", "float64"] and n_unics < 4:
            # numèrica però amb pocs valors únics: categórica (0,1)
            recompte = serie.value_counts().sort_index()                #El sort_index m'ordena els valors de major a menor.
            recompte.plot(kind="bar")
            plt.title(f"{nom_columna} (Categorica - binaria -)")        # A la gràfica li posem com a títol el nom de la columna i entre parentesis que és Cdel tipus CATEGORICA.
        elif serie.dtype in ["int64", "float64"]:
            # numérica “real”
            if len(df) > 10:
                serie.plot.box()
                plt.title(f"{nom_columna} - boxplot")
            else:
                plt.hist(serie.dropna(), bins=10, edgecolor="black")       #bins=10 ens diu que són valors binaris fins a 10. 
                plt.title(f"{nom_columna} - histograma")
        else:
            # texto/categórica
            conteo = serie.value_counts().head(10)
            conteo.plot(kind="bar")
            plt.title(f"{nom_columna} - més freqüents")
            plt.xticks(rotation=45)

        plt.ylabel("Quantitat")
        plt.tight_layout()

        # usar SOLO el nombre de la columna para el fichero
        nombre_fichero = f"{str(nom_columna)}.png"
        plt.savefig(carpeta / nombre_fichero)
        plt.close()

    print(f"Okay, ja ho tenim això: {len(df.columns)} gràfics en la carpeta 'Gràfics_automatitzats'")

```

<span style="font-size:14px; color:salmon ">**(EXTRA)** Des de quina ciutat la ruta seria més curta amb l'algoritme plantejat

```python
def info_distancies(matriu, origen):
    if origen not in matriu.index:
        raise ValueError(f"La ciutat '{origen}' no existeix a la matriu.")
    dist_origen = matriu[origen]
    dist_ordenada = dist_origen.sort_values(na_position='first')
    diferencia = dist_ordenada.diff()
    df = pd.DataFrame({
        "Distància des d'origen": dist_ordenada,
        "Diferència": diferencia
    })

    return df
```

```python
ciutat=input("Entra una ciutat d'origen: ")
df_resultat = info_distancies(mtrx_neta, ciutat)
df_resultat
```

<span style="color:peru; font-size:14px">**Què pasa si l'usuari vol comparar més d'una ciutat?**

<span style="font-size:12px">Donada una matriu de distàncies i una llista de ciutats d'origen, retorna un DataFrame amb:
* <span style="font-size:12px">Ciutats com a index
* <span style="font-size:12px">Per cada ciutat d'origen:
    * <span style="font-size:12px">Distància ordenada
    * <span style="font-size:12px">Diferència 

```python
def info_distancies_multi(matriu, ciutats_origen):
    for ciutat in ciutats_origen:
        if ciutat not in matriu:
            raise ValueError(f"La ciutat '{ciutat}' no existeix a la matriu.")

    # DataFrame final buit
    df_final = pd.DataFrame(index=matriu.index)

    # Per cada ciutat d'origen, calculem distàncies i diferències
    for origen in ciutats_origen:
        dist_origen = matriu[origen]

        # Ordenar distàncies
        dist_ordenada = dist_origen.sort_values(na_position='first')

        diferencia = dist_ordenada.diff()
        df_final[f"Distància des de {origen}"] = dist_ordenada
        #df_final[f"Diferència {origen}"] = diferencia

    return df_final
```

```python
ciutat = input("Entra ciutat o ciutats d'origen (separades per comes): ")
llista_ciutats = [c.strip() for c in ciutat.split(",")]

recor = info_distancies_multi(mtrx_neta, llista_ciutats)
recor

#REVISAR OUTPUT perqùe NO s'estan càlculant bé les dades ja que aquestes no es poden ordenar bé perquè estem seleccionant múltiples
# amb el mateix index...
#NO ordena bé les ciutats perquè posa en l'index en les primeres posicions les ciutats que hem introdeuit en la llista
```
